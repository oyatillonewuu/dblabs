```python
import matplotlib.pyplot as plt
import pandas as pd
```

```python
# Global vars

columns = ["sepallength", "sepalwidth", "petallength", "petalwidth", "class"]
```

```python
# Load Dataset
url = "https://raw.githubusercontent.com/jbrownlee/Datasets/master/iris.csv"
df = pd.read_csv(url, names=columns)
```

```python
# Intialize the setup
fig, (chart1, chart2) = plt.subplots(nrows=1, ncols=2, figsize=(12, 5))

# Chart 1 (Bar): Average sepal length per flower class

# Googled this
# Can be checked for correctness using the custom code in comment in the last code section
classes = list(set(df["class"]))
class_sepallen_mean_df = df.groupby("class")["sepallength"].mean().to_frame()

class_mean_pairs = list(class_sepallen_mean_df.to_dict()["sepallength"].items())
classes = []
means = []

for flower_class, mean in class_mean_pairs:
  classes.append(flower_class)
  means.append(mean)

chart1.bar(
    x=classes,
    height=means,
    color="orange"
  )

chart1.set_title("Mean sepallength by classes")
chart1.set_xlabel("Class")
chart1.set_ylabel("Mean length (cm)")

# Chart 2 (Histogram): Distribution of petal length across all flowers

petallengths_df = df["petallength"]

chart2.hist(x=petallengths_df, bins=20, color="skyblue")
chart2.set_title("Petallength distribution across all flowers")
chart2.set_xlabel("Length (cm)")
chart2.set_ylabel("Frequency")
plt.tight_layout()
plt.show()
```
